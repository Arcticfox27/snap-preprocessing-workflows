```python
import time
import subprocess
import os
from osgeo import gdal
```

```python
Input_Directory  = 'E:/Data/2_Data/Mafra_2020/ALOS2/'
Output_Directory = 'E:/Data/3_Data/Mafra_2020/ALOS2/'
```

```python
# ALOS-2 Dictionaries
GLCM_BETA       = ['GLCMBETAHHASM', 'GLCMBETAHHCONTRAST', 'GLCMBETAHHDISSIMILARITY', 'GLCMBETAHHENERGY', 'GLCMBETAHHENTROPY', 'GLCMBETAHHCORRELATION', 'GLCMBETAHHMEAN', 'GLCMBETAHHVARIANCE', 'GLCMBETAHHHOMOGENEITY', 'GLCMBETAHHMAX',
                   'GLCMBETAHVASM', 'GLCMBETAHVCONTRAST', 'GLCMBETAHVDISSIMILARITY', 'GLCMBETAHVENERGY', 'GLCMBETAHVENTROPY', 'GLCMBETAHVCORRELATION', 'GLCMBETAHVMEAN', 'GLCMBETAHVVARIANCE', 'GLCMBETAHVHOMOGENEITY', 'GLCMBETAHVMAX']
GLCM_SIGMA      = ['GLCMSIGMAHHASM', 'GLCMSIGMAHHCONTRAST', 'GLCMSIGMAHHDISSIMILARITY', 'GLCMSIGMAHHENERGY', 'GLCMSIGMAHHENTROPY', 'GLCMSIGMAHHCORRELATION', 'GLCMSIGMAHHMEAN', 'GLCMSIGMAHHVARIANCE', 'GLCMSIGMAHHHOMOGENEITY', 'GLCMSIGMAHHMAX',
                   'GLCMSIGMAHVASM', 'GLCMSIGMAHVCONTRAST', 'GLCMSIGMAHVDISSIMILARITY', 'GLCMSIGMAHVENERGY', 'GLCMSIGMAHVENTROPY', 'GLCMSIGMAHVCORRELATION', 'GLCMSIGMAHVMEAN', 'GLCMSIGMAHVVARIANCE', 'GLCMSIGMAHVHOMOGENEITY', 'GLCMSIGMAHVMAX']
GLCM_GAMMA      = ['REMOVE', 'GLCMGAMMAHHASM', 'GLCMGAMMAHHCONTRAST', 'GLCMGAMMAHHDISSIMILARITY', 'GLCMGAMMAHHENERGY', 'GLCMGAMMAHHENTROPY', 'GLCMGAMMAHHCORRELATION', 'GLCMGAMMAHHMEAN', 'GLCMGAMMAHHVARIANCE', 'GLCMGAMMAHHHOMOGENEITY', 'GLCMGAMMAHHMAX',
                             'GLCMGAMMAHVASM', 'GLCMGAMMAHVCONTRAST', 'GLCMGAMMAHVDISSIMILARITY', 'GLCMGAMMAHVENERGY', 'GLCMGAMMAHVENTROPY', 'GLCMGAMMAHVCORRELATION', 'GLCMGAMMAHVMEAN', 'GLCMGAMMAHVVARIANCE', 'GLCMGAMMAHVHOMOGENEITY', 'GLCMGAMMAHVMAX']
LIA             = ['LIA']
RATIOS          = ['BETAND', 'BETARATIO', 'GAMMAND', 'GAMMARATIO', 'SIGMAND', 'SIGMARATIO']
SCAT            = ['BETAHH', 'BETAHV', 'GAMMAHH', 'GAMMAHV', 'SIGMAHH', 'SIGMAHV']
```

```python
# Count time
Time_Process_Start = time.perf_counter()
# Get Raster files
Rasters = [Input_Directory + Path for Path in os.listdir(Input_Directory)]
# Iterate Rasters to split them
for Raster in Rasters:
    # Get Raster Name and Type
    Raster_Name = Raster.split('/')[-1].split('.')[0]
    Raster_Type = Raster_Name.split('_')[3]
    # Get Raster bands count
    DS               = gdal.Open(Raster)
    Raster_BandCount = DS.RasterCount
    DS = None
    # Print which raster is being processed
    print(Raster_Name)
    # Select the Correct dictionary 
    if Raster_Type == 'GLCMBETA':
        Bands_Names = GLCM_BETA
    elif Raster_Type == 'GLCMSIGMA':
        Bands_Names = GLCM_SIGMA     
    elif Raster_Type == 'GLCMGAMMA':
        Bands_Names = GLCM_GAMMA
    elif Raster_Type == 'LIA':
        Bands_Names = LIA
    elif Raster_Type == 'RATIOS':
        Bands_Names = RATIOS
    elif Raster_Type == 'SCAT':
        Bands_Names = SCAT
    # Iterate Bands for Splitting
    for i in range(1, Raster_BandCount + 1):       
        Output_File_Path = Output_Directory + Raster_Name.replace(Raster_Type, Bands_Names[i-1]) + '.tif'
        if 'REMOVE' in Bands_Names[i-1]:
            continue
        Command = 'gdal_translate -b ' + str(i) + ' -tr 25 25 -r cubic ' + Raster + ' ' + Output_File_Path
        subprocess.run(Command, check=True)
    print('----')
# Compute processing time
Time_Process_Ends = time.perf_counter()
Processing_Time   = Time_Process_Ends - Time_Process_Start
# Convert to h:m:s
hours, rem       = divmod(Processing_Time, 3600)
minutes, seconds = divmod(rem, 60)
print(f"Elapsed time: {int(hours):02d}:{int(minutes):02d}:{seconds:05.2f}")
```
