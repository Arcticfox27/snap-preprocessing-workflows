```python
import time
import subprocess
import os
from osgeo import gdal
```

```python
Input_Directory  = 'C:/Users/jepir/Desktop/S1-GRD/Input/'
Output_Directory = 'C:/Users/jepir/Desktop/S1-GRD/Output/'
```

```python
# ALOS-2 Dictionaries
BETA       = ['BETAVHCONTRAST', 'BETAVHDISSIMILARITY', 'BETAVHHOMOGENEITY', 'BETAVHASM', 'BETAVHENERGY', 'BETAVHMAX', 'BETAVHENTROPY', 'BETAVHMEAN', 'BETAVHVARIANCE', 'BETAVHCORRELATION',
              'BETAVVCONTRAST', 'BETAVVDISSIMILARITY', 'BETAVVHOMOGENEITY', 'BETAVVASM', 'BETAVVENERGY', 'BETAVVMAX', 'BETAVVENTROPY', 'BETAVVMEAN', 'BETAVVVARIANCE', 'BETAVVCORRELATION',
              'BETAVH', 'BETAVV']


GAMMASIGMA = ['GAMMAVHCONTRAST', 'GAMMAVHDISSIMILARITY', 'GAMMAVHHOMOGENEITY', 'GAMMAVHASM', 'GAMMAVHENERGY', 'GAMMAVHMAX', 'GAMMAVHENTROPY', 'GAMMAVHMEAN', 'GAMMAVHVARIANCE', 'GAMMAVHCORRELATION',
              'GAMMAVVCONTRAST', 'GAMMAVVDISSIMILARITY', 'GAMMAVVHOMOGENEITY', 'GAMMAVVASM', 'GAMMAVVENERGY', 'GAMMAVVMAX', 'GAMMAVVENTROPY', 'GAMMAVVMEAN', 'GAMMAVVVARIANCE', 'GAMMAVVCORRELATION',
              'SIGMAVHCONTRAST', 'SIGMAVHDISSIMILARITY', 'SIGMAVHHOMOGENEITY', 'SIGMAVHASM', 'SIGMAVHENERGY', 'SIGMAVHMAX', 'SIGMAVHENTROPY', 'SIGMAVHMEAN', 'SIGMAVHVARIANCE', 'SIGMAVHCORRELATION',
              'SIGMAVVCONTRAST', 'SIGMAVVDISSIMILARITY', 'SIGMAVVHOMOGENEITY', 'SIGMAVVASM', 'SIGMAVVENERGY', 'SIGMAVVMAX', 'SIGMAVVENTROPY', 'SIGMAVVMEAN', 'SIGMAVVVARIANCE', 'SIGMAVVCORRELATION',
              'GAMMAVH', 'GAMMAVV', 'SIGMAVH', 'SIGMAVV', 'LIA', 'DEM']
```

```python
# Count time
Time_Process_Start = time.perf_counter()
# Get Raster files
Rasters = [Input_Directory + Path for Path in os.listdir(Input_Directory)]
# Iterate Rasters to split them
for Raster in Rasters:
    # Get Raster Name and Type
    Raster_Name = Raster.split('/')[-1].split('.')[0]
    Raster_Type = Raster_Name.split('_')[2]
    # Get Raster bands count
    DS               = gdal.Open(Raster)
    Raster_BandCount = DS.RasterCount
    DS = None
    # Print which raster is being processed
    print(Raster_Name)
    # Select the Correct dictionary 
    if Raster_Type == 'Beta':
        Bands_Names = BETA
    elif Raster_Type == 'GammaSigma':
        Bands_Names = GAMMASIGMA     
    # Iterate Bands for Splitting
    for i in range(1, Raster_BandCount + 1):       
        Output_File_Path = Output_Directory + Raster_Name.replace(Raster_Type, Bands_Names[i-1]) + '.tif'
        if 'REMOVE' in Bands_Names[i-1]:
            continue
        Command = 'gdal_translate -b ' + str(i) + ' ' + Raster + ' ' + Output_File_Path
        subprocess.run(Command, check=True)
        print(Command)
    print('----')
# Compute processing time
Time_Process_Ends = time.perf_counter()
Processing_Time   = Time_Process_Ends - Time_Process_Start
# Convert to h:m:s
hours, rem       = divmod(Processing_Time, 3600)
minutes, seconds = divmod(rem, 60)
print(f"Elapsed time: {int(hours):02d}:{int(minutes):02d}:{seconds:05.2f}")
```
