```python
# Get Files Paths
Files_Paths = [
    os.path.join(Input_Directory, Directory, Path)
    for Directory in os.listdir(Input_Directory)
    for Path in os.listdir(os.path.join(Input_Directory, Directory))
]
```

```python
Files_Paths
```

```python
# Count time
Time_Process_Start = time.perf_counter()
for Path in Files_Paths:
    Raster_SET  = Path.split('\\')[-1].split('_')[2]
    Raster_TYPE = Path.split('\\')[-1].split('_')[3]
    # Define Command Based on Raster Set/Type
    if Path.endswith('.tif'):
        print(Raster_SET + '_' + Raster_TYPE)
        if Raster_SET == 'ALOS2':  
            Command = 'gdal_translate -a_nodata -9999 ' + Path + ' ' + Path.replace('3_Data', '4_Data')
            print(1)
        elif Raster_SET == 'S2':
            Command = 'gdal_translate -a_nodata -9999 ' + Path + ' ' + Path.replace('3_Data', '4_Data')
            print(2)
        elif Raster_SET == 'S1SLC':
            if 'GLCM' in Raster_TYPE:
                Command = 'gdal_translate -a_nodata ' + str(np.nan) + ' ' + Path + ' ' + Path.replace('3_Data', '4_Data')
                print(3)
            else:
                Command = 'gdal_translate -a_nodata 0 ' + Path + ' ' + Path.replace('3_Data', '4_Data')
                print(4)
        else:
            if any(sub in Raster_TYPE for sub in GRD_GLCM):
                Command = 'gdal_translate -a_nodata -9999 ' + Path + ' ' + Path.replace('3_Data', '4_Data')
                print(5)
            else:
                Command = 'gdal_translate -a_nodata 0 ' + Path + ' ' + Path.replace('3_Data', '4_Data')
                print(6)
        subprocess.run(Command, check=True)
        print('----')
# Compute processing time
Time_Process_Ends = time.perf_counter()
Processing_Time   = Time_Process_Ends - Time_Process_Start
# Convert to h:m:s
hours, rem       = divmod(Processing_Time, 3600)
minutes, seconds = divmod(rem, 60)
print(f"Elapsed time: {int(hours):02d}:{int(minutes):02d}:{seconds:05.2f}")
```
