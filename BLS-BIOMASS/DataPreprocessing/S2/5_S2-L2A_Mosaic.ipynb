```python
import time
import subprocess
import os
from osgeo import gdal
import numpy as np
```

```python
Input_Directory  = 'C:/Users/jepir/Desktop/S1-SLC/Singleband/Extremadura_20170000/'
Output_Directory = 'C:/Users/jepir/Desktop/S1-SLC/Singleband/Extremadura_20170000/Mosaic/'
```

```python
Files_Paths = [[Path for Path in os.listdir(os.path.join(Input_Directory, Directory)) if Path.endswith('.tif')] for Directory in os.listdir(Input_Directory)]
# Get Number of Bands and Tiles
N_Bands = len(Files_Paths[0])
N_Mosaics = len(Files_Paths)
```

```python
# Count time
Time_Process_Start = time.perf_counter()
for Path in Files_Paths[0]:
    Time_Process_Start_Command = time.perf_counter()
    Path_Code = Path.split('_')[3]
    Input_Rasters = [[os.path.join(Input_Directory,Input_Path.split('_')[-1].split('.')[0],Input_Path) for Input_Path in List if Input_Path.split('_')[3] == Path_Code] for List in Files_Paths]
    # Convert in array of arrays
    Input_Rasters = [Item for List in Input_Rasters for Item in List]
    #Output_Raster = os.path.join(Output_Directory, Path.replace('_A.tif', '_Mosaic.tif'))
    Output_Raster = os.path.join(Output_Directory, Path)
    print(Input_Rasters)
    # Execute Mosaicking comand
    cmd = [
        "gdalwarp",
        "-r", "average",                    # Average overlapping pixels
        "-wo", "NUM_THREADS=ALL_CPUS",      # Use all CPU cores
        "-of", "GTiff",                     # Output format
        "-srcnodata", "-9999",                 # Treat NaN as NoData in input rasters
        "-dstnodata", "nan",                # Output NoData will also be NaN
        "-cblend", "10",
        *Input_Rasters,
        Output_Raster
    ] 
    # Run the command
    subprocess.run(cmd, check=True)
    print('-+-+-+-+-+-+-')
    # Compute processing time
    Time_Process_Ends = time.perf_counter()
    Processing_Time   = Time_Process_Ends - Time_Process_Start_Command
    # Convert to h:m:s
    hours, rem       = divmod(Processing_Time, 3600)
    minutes, seconds = divmod(rem, 60)
    print(f"Elapsed time: {int(hours):02d}:{int(minutes):02d}:{seconds:05.2f}")
# Compute processing time
Time_Process_Ends = time.perf_counter()
Processing_Time   = Time_Process_Ends - Time_Process_Start
# Convert to h:m:s
hours, rem       = divmod(Processing_Time, 3600)
minutes, seconds = divmod(rem, 60)
print(f"Elapsed time: {int(hours):02d}:{int(minutes):02d}:{seconds:05.2f}")
```
